# PathSnap Phase 3: Machine Learning Model Training, Evaluation and Comparison

2. **Support Vector Machine (LinearSVC with Probability Calibration)**

4. **XGBoost**
5. **Multi-Layer Perceptron (MLP)**

```python
import os, sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json

# Add project root to sys.path
PROJECT_ROOT = os.path.abspath('..')
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.models.data_loader import load_config, dataset_audit
from src.models.evaluate_models import build_comparison_table

config = load_config(os.path.join(PROJECT_ROOT, 'configs', 'phase3_config.yaml'))
print('Phase 3 Config loaded.')
```

## 1. Dataset Audit and Trip-Level Split

```python
audit = dataset_audit(config)
print('Train trips:', audit['split']['n_train_trips'])
print('Val trips:', audit['split']['n_val_trips'])
print('Trip overlap count:', audit['split']['overlap_count'])
print('Train positive rate:', f"{audit['target']['train']['pct_pos']}%")
print('Val positive rate:', f"{audit['target']['val']['pct_pos']}%")
print('Train candidate coverage:', f"{audit['target']['train']['coverage_pct']}%")
print('Val candidate coverage:', f"{audit['target']['val']['coverage_pct']}%")
```
